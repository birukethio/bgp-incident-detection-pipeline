# 08 - MRT Pilot Download

Controlled raw-data acquisition step only. This notebook summarizes the refined 9-incident TRAIN pilot download using candidate active segments for Hijack/Leak and documented Cloudflare intervals for Outage. It does not parse MRT data, generate feature windows, create Normal labels, add CAIDA/RPKI data, train models, or touch the frozen roster/partitions.

```python
from pathlib import Path
import csv
import sys
from collections import Counter

ROOT = Path.cwd().resolve()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent.resolve()
sys.path.insert(0, str(ROOT / 'src'))

def read_csv(path):
    with path.open(newline='', encoding='utf-8') as handle:
        return list(csv.DictReader(handle))
```

## Refined Download Manifest Summary

```python
manifest = read_csv(ROOT / 'data/pilot/mrt_pilot_download_manifest_v1.csv')
print('Planned files:', len(manifest))
print('Class counts:', Counter(row['category'] for row in manifest))
print('Collectors:', Counter(row['routeviews_collector'] for row in manifest))
print('Expected status:', Counter(row['expected_status'] for row in manifest))
```

## Download Progress and File Status

```python
file_status = read_csv(ROOT / 'data/pilot/mrt_pilot_download_file_status_v1.csv')
print('File status rows:', len(file_status))
print('Downloaded:', Counter(row['downloaded'] for row in file_status))
print('HTTP status:', Counter(row['http_status'] for row in file_status))
print('Total bytes:', sum(int(row['file_size_bytes']) for row in file_status))
```
