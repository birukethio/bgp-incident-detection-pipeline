# 12 - Normal Baseline Design Pilot

Controlled TRAIN-only design step for Normal baseline qualification. This notebook reviews candidate control periods, exclusion-buffer sensitivity, sustained-stability feasibility, behavioral schema freeze, and dry-run RouteViews acquisition planning.

It does not download Normal MRT files, create `NORMAL_VERIFIED` labels, train models, add CAIDA/RPKI, or alter frozen roster/partition files.

```python
from pathlib import Path
import pandas as pd

ROOT = Path('..').resolve()
PILOT = ROOT / 'data' / 'pilot'

schema = pd.read_csv(PILOT / 'behavioral_feature_schema_freeze_v1.csv')
candidates = pd.read_csv(PILOT / 'normal_candidate_periods_v1.csv')
buffer = pd.read_csv(PILOT / 'normal_exclusion_buffer_sensitivity_v1.csv')
stability = pd.read_csv(PILOT / 'normal_sustained_stability_sensitivity_v1.csv')
summary = pd.read_csv(PILOT / 'normal_baseline_candidate_summary_v1.csv')
plan = pd.read_csv(PILOT / 'normal_mrt_acquisition_plan_v1.csv')
```
