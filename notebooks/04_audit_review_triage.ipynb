## Simple Visualization

```python
import matplotlib.pyplot as plt

rows = pd.concat([
    pd.read_csv(PARTITION_DIR / "train_incidents.csv"),
    pd.read_csv(PARTITION_DIR / "validation_incidents.csv"),
    pd.read_csv(PARTITION_DIR / "holdout_incidents.csv"),
], ignore_index=True)
rows["duration_minutes"] = pd.to_numeric(rows["duration_minutes"], errors="coerce")
rows.boxplot(column="duration_minutes", by="category", showfliers=False, figsize=(8, 4))
plt.title("Duration by class, outliers hidden")
plt.suptitle("")
plt.ylabel("duration_minutes")
plt.tight_layout()
```
